```python
class equation:
  def __init__(self, a,b ,c):
    self.a = a
    self.b = b
    self.c = c

  def __str__(self):
      return f"{self.a}x+{self.b}y+{self.c}=0"

  def intersect_of_equation(self,other):
    # Check for coincident lines (many intersections)
    if (self.a * other.b == self.b * other.a) and \
       (self.b * other.c == self.c * other.b) and \
       (self.a * other.c == self.c * other.a):
        return "Many intersections (coincident lines)"

    # Check for parallel lines (no intersection)
    elif (self.a * other.b == self.b * other.a) and \
         (self.b * other.c != self.c * other.b):
        return "No intersection (parallel lines)"

    # All other cases result in one intersection (intersecting lines)
    else:
        return "One intersection (intersecting lines)"

b=equation(1,4,8)
b1 = equation(1,6,-9)

print(f"Equation 1: {b}")
print(f"Equation 2: {b1}")
print(f"Intersection of Equation 1 and Equation 2: {b.intersect_of_equation(b1)}")

```

```python
class Point:
   def __init__(self,a,b):
    self.cod_a = a
    self.cod_b = b

   def __str__(self):
      return f"<{self.cod_a},{self.cod_b}"

   def euclid_distance(self, other):
    return ((self.cod_a-other.cod_a)**2+(self.cod_b-other.cod_b)**2)**0.5


   def distance_from_origin(self):

    return ((self.euclid_distance(Point(0, 0))))



p1 = Point(8,4)
p2 = Point(5,6)
print(p1.euclid_distance(p2))

p1.distance_from_origin()
```

```python
class prime():
  def __init__(self, num):
    self.num = num

  def is_prime(self):
    if self.num<2:
      return False
    else:
      for i in range(2, self.num):
        if self.num%i==0:
          return False

      return True

object = prime(10)
print(object.is_prime())
```

```python
class Matrix():
  def __init__(self,data):
    self.data = data

  def __str__(self):
      return str(self.data)
  def __add__(self, other):
    new_data = []

    for i in range(len(self.data)):
      row =[]
      for j in range (len(self.data[0])):
        row.append(self.data[i][j]+ other.data[i][j])

      new_data.append(row)

    return Matrix(new_data)

  def __sub__(self, other):
    new_data = []
    for i in range(len(self.data)):
      row = []
      for j in range(len(self.data[0])):
        row.append(self.data[i][j]-other.data[i][j])
      new_data.append(row)
    return Matrix(new_data)

  def __mul__(self, other):
    new_data = []
    for i in range(len(self.data)):
      row = []
      for j in range(len(self.data[0])):
        row.append(self.data[i][j]-other.data[i][j])
      new_data.append(row)
    return Matrix(new_data)


a = Matrix([[1, 2, 3], [4, 5, 6, ]])
b = Matrix([[7, 8, 9], [10, 11, 12]])
print(a+b)
print(a-b)


```

```python
class fraction:
  def __init__(self, a, b):
    self.num = a
    self.demo = b

  def __str__(self):
    return f"{self.num}/{self.demo}"


  def __add__(self, other):
    new_num = self.num*other.demo+ self.demo*other.num
    new_demo = self.demo*other.demo

    return fraction(new_num, new_demo)

  def __sub__(self, other):
    new_num = self.num*other.demo- self.demo*other.num
    new_demo = self.demo*other.demo

    return fraction(new_num, new_demo)

  def __mul__(self, other):
    new_num = self.num*other.num
    new_demo = self.demo*other.demo

    return fraction(new_num, new_demo)

  def __truediv__(self, other):
    new_num = self.num*other.demo
    new_demo = self.demo*other.num

    return fraction(new_num, new_demo)

object=fraction(3,4)
object1=fraction(5,6)
print(object+object1)
print(object-object1)
print(object/object1)
```

```python

class Atm:

  def __init__(self):
    self.pin = ""
    self.balance = 0
    self.menu()

  def menu(self):
    user_input = input("""
    press 1 for create pin
    press 2 for check blance
    press 3 for pin reset
    press 4 for deposit
    press 5 for withdraw
    press 6 for exit
    """)
    if user_input == "1":
      self.create_pin()
    elif user_input == "2":
      self.check_balance()
    elif user_input == "3":
      self.pin_reset()
    elif user_input == "4":
      self.deposite()
    elif user_input =="5":
      self.withdraw()
      exit()
    else:
      print("Invalid option. Please try again.")
      self.menu()

  def create_pin(self):
    self.pin = input("enter your pin: ")
    print("pin set successfully")
    self.menu()

  def check_balance(self):
    pin = input("enter your pin: ")
    if pin == self.pin:
      print("your balance is",self.balance)
    else:
      print("Invalid pin")
    self.menu()

  def pin_reset(self):
    old_pin = input("Enter your old pin: ")
    if old_pin == self.pin:
      new_pin = input("Enter your new pin: ")
      self.pin = new_pin
      print("Pin reset successfully!")
    else:
      print("Incorrect old pin.")
    self.menu()

  def deposite(self):
    pin = input("Enter your pin: ")
    if pin == self.pin:
      amount = float(input("Enter amount to deposit: "))
      if amount > 0:
        self.balance += amount
        print(f"Amount {amount} deposited successfully. New balance: {self.balance}")
      else:
        print("Deposit amount must be positive.")
    else:
      print("Invalid pin.")
    self.menu()

  def withdraw(self):
    pin = input("Enter your pin: ")
    if pin == self.pin:
      amount = float(input("Enter amount to withdraw: "))
      if 0 < amount <= self.balance:
        self.balance -= amount
        print(f"Amount {amount} withdrawn successfully. New balance: {self.balance}")
      elif amount <= 0:
        print("Withdrawal amount must be positive.")
      else:
        print("Insufficient balance.")
    else:
      print("Invalid pin.")
    self.menu()


object=Atm()
```
